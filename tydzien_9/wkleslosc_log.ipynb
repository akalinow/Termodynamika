```python
import matplotlib.pyplot as plt
import numpy as np

#Increase plots font size
params = {'legend.fontsize': 'xx-large',
          'figure.figsize': (5, 5),
         'axes.labelsize': 'xx-large',
         'axes.titlesize':'xx-large',
         'xtick.labelsize':'xx-large',
         'ytick.labelsize':'xx-large'}
plt.rcParams.update(params)
```

```python
x = np.linspace(1, 2, 100)
y = np.log(x)

x1 = [x[0], x[-1]]
y1 = [y[0], y[-1]]

plt.plot(x, y, label=r'$log(f_a\frac{V_a}{n_a} + f_b\frac{V_b}{n_b})$')
plt.plot(x1, y1, "r--", label=r'$f_a\log(\frac{V_a}{n_a})+ f_b\log(\frac{V_b}{n_b})$')
plt.plot(x1[0], y1[0], 'ro', label=r'$\log(\frac{V_a}{n_a}),~f_a=1,~f_b=0$')
plt.plot(x1[1], y1[1], 'go', label=r'$\log(\frac{V_b}{n_b}),~f_a=0,~f_b=1$')
plt.xlabel(r'$f_a\frac{V_a}{n_a} + f_b\frac{V_b}{n_b}$')
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5));
```
